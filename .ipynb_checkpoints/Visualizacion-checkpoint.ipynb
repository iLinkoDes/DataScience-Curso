```python
import bokeh.plotting as bk

bk.output_notebook()
```

```python
df.Indice_Criminalidad
```

```python
df["Crimen_Quintil"] = pd.qcut(df.Indice_Criminalidad,5)
```

```python
df.Crimen_Quintil.cat.categories
```

```python
from bokeh.palettes import brewer

colors = brewer["Spectral"][len(df.Crimen_Quintil.unique())]
colors
```

```python
p = bk.figure(
    plot_width=600,
    plot_height=600,
    title="Habitaciones vs Valor Vivienda vs Crimen"
)

for i, quintil in enumerate(df.Crimen_Quintil.cat.categories):
    df_q = df[df.Crimen_Quintil==quintil]
    p.scatter(df_q.N_Habitaciones_Medio, df_q.Valor_Mediano, color=colors[i], legend_label="({} - {})".format(quintil.left,quintil.right))
    
bk.show(p)
```

```python
hist, edges = np.histogram(df.Valor_Mediano, bins=20)
hist
```

```python
edges
```

```python
pl = bk.figure(title="Histograma valor viviendas", tools="save,hover", background_fill_color="#E8DDCB")

pl.quad(top=hist, bottom=0, left=edges[:-1], right=edges[1:], fill_color="#026560")
bk.show(pl)
```
